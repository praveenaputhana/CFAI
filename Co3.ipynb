```python
analysts = ["A1", "A2", "A3"]

shifts = ["Morning", "Afternoon", "Night"]

assignment = {}

def assign_shift(index):

    if index == len(shifts):
        return True

    shift = shifts[index]

    for analyst in analysts:

        if analyst not in assignment.values():

            assignment[shift] = analyst

            if assign_shift(index + 1):
                return True

            del assignment[shift]

    return False


if assign_shift(0):

    print("Shift Assignment:\n")

    for shift, analyst in assignment.items():
        print(shift, "->", analyst)

else:
    print("No valid assignment found")
```
