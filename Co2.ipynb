```python
from collections import deque

response_graph = {
    "Incident": ["Alert"],
    "Alert": ["Block_IP", "Increase_Monitoring"],
    "Block_IP": [],
    "Increase_Monitoring": ["Notify_Admin"],
    "Notify_Admin": []
}

def bfs(start):

    visited = set()
    queue = deque([start])

    print("Response Plan:")

    while queue:
        node = queue.popleft()

        if node not in visited:
            print(node)
            visited.add(node)

            for neighbor in response_graph[node]:
                queue.append(neighbor)

# Driver

bfs("Incident")
```
