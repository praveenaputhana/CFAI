```python
class IncidentAgent:

    def __init__(self):
        self.rules = {
            "failed_logins": 5,
            "suspicious_traffic": 1000
        }

    def detect_incident(self, failed_logins, traffic):

        if failed_logins > self.rules["failed_logins"]:
            return "Possible Brute Force Attack"

        elif traffic > self.rules["suspicious_traffic"]:
            return "Suspicious Network Activity"

        else:
            return "System Normal"


# Driver Code

agent = IncidentAgent()

failed_logins = int(input("Enter failed login attempts: "))
traffic = int(input("Enter suspicious traffic count: "))

result = agent.detect_incident(failed_logins, traffic)

print("\nIncident Status:", result)
```
